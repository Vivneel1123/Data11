```python
# 23.	Filter even numbers from nums = list(range(10)) into a new list using a loop and conditional check. Print both lists.
# Filter even numbers.
nums = list(range(10))

even_nums = []

for n in nums:
    if n % 2 == 0:
        even_nums.append(n)

print(nums)
print(even_nums)
```

```python
# 26.	Simulate data cleaning: Remove duplicates from raw_data = [1, 2, 2, 3, 1, 4] using a loop or set conversion (back to list). Print unique list.
# Remove duplicates.
raw_data = [1, 2, 2, 3, 1, 4]
unique_list = []

for item in raw_data:
    if item not in unique_list:
        unique_list.append(item)

print(unique_list)
```

```python
# 27.	Given customer ratings ratings = [4.5, 3.0, 5.0, 2.5], slice and print ratings above 4.0 using indexing/loop.
# Ratings above 4.0.
ratings = [4.5, 3.0, 5.0, 2.5]

high = []

for r in ratings:
    if r > 4.0:
        high.append(r)

print(high)
```

```python
# 27.	Given customer ratings ratings = [4.5, 3.0, 5.0, 2.5], slice and print ratings above 4.0 using indexing/loop.
# Ratings above 4.0.
ratings = [4.5, 3.0, 5.0, 2.5]

# Slice and print ratings above 4.0
high_ratings = [r for r in ratings if r > 4.0]
print(high_ratings)
```

```python
# 29.	Analyze a dataset: heights = [160, 175, 168, 182]; add new height 170, sort, and print average using basic arithmetic on list.
# Analyze heights dataset.
heights = [160, 175, 168, 182]

heights.append(170)

heights.sort()

average = sum(heights) / len(heights)

print(heights)
print('Average:', average)
```
